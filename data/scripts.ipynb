```python
import pandas as pd

df = pd.read_json("rapport.json")

for metric in ["cpu_usage", "memory_usage", "latency_ms", "disk_usage", "error_rate", "io_wait", "temperature_celsius","power_consumption_watts"]:
    print(
          f"p75={df[metric].quantile(0.75):.2f}, p90={df[metric].quantile(0.9):.2f}, "
          f"p99={df[metric].quantile(0.99):.2f}, max={df[metric].max():.2f}")
```
